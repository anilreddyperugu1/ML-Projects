# **Market Basket Analysis using ECLAT**

**Importing Dependencies**

```python
import numpy as np
import pandas as pd
from apyori import apriori
```

**---- INITIAL DATA ANALYSIS & PRE PROCESSING ----**

```python
dataset = pd.read_csv('Groceries_dataset.csv') #reading the dataset
```

```python
dataset.head() #gives top 5 rows
```

**Creating transactions needed for ECLAT analysis**

```python
transactions = dataset.groupby(['Member_number', 'Date'])['itemDescription'].apply(list).values.tolist() 
```

**Analysis of the above operation**

```python
for i in range(5):
    print(transactions[i])
```

**Definig rules and training the model**

```python
frequent_itemsets = apriori(transactions, min_support=0.001, min_confidence=0.1, min_lift=0.5, min_length=2) #defining thresholds
```

**Model Results**

```python
rules = list(frequent_itemsets) #output
for i, (rule) in enumerate(rules):
    if i == 50:
        break
    print(rule)
```

**Converting the results into DataFrame**

```python
records = []
for rule in rules:
    for stat in rule.ordered_statistics:
        if len(stat.items_base) == 0:  # skip single-item rules
            continue
        records.append({
            'lhs': tuple(stat.items_base)[0],
            'rhs': tuple(stat.items_add)[0],
            'support': rule.support,
        })

df_rules = pd.DataFrame(records)
df_rules.head(15)
```
